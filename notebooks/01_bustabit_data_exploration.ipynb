# Bustabit Data Exploration

Bustabit is an online gambling website that collects data on its actual player base. 

### Gameplay

    Players make bets. There is a multiplier that starts at 1.00x and increases (1.10, 1.50, 2.00, 3.00, etc). The player can click "cash out" to lock in the desired multiplier. The winnings from your bet are multiplied by the same multiplier the player "cashed out" at. 

    There is a bust mechanic to the game, where in any moment the game is being played there is a chance the multiplier stops increasing and does a bust to 0. All players that have not cashed out before that happen lose their bet. 

    The longer the player waits to cash out, the higher the multiplier gets, but also the higher the risk of losing. 

#### The following is an initial look at the raw dataset: structure, summary statistics, and a sample of rows.

```python
import pandas as pd

from poker_coach.utils import rainbow
from poker_coach.config import ID_COLS, BUSTABIT_DATA_FILE, BITS_PER_BTC, BTC_USD_RATE
```

```python
df = pd.read_csv(BUSTABIT_DATA_FILE, parse_dates=['PlayDate'])
df['Won'] = df['CashedOut'].notna()   # missing CashedOut = player busted (lost)

# Bet/Bonus/Profit are recorded in bits (1 bit = 0.000001 BTC), convert to BTC so the values read as real money
for col in ['Bet', 'Bonus', 'Profit']:
    df[col] = df[col] / BITS_PER_BTC

print(f"{len(df)} bets from {df['Username'].nunique()} players")
rainbow(df.drop(columns=ID_COLS).describe())
```

Identical version of the table above, converted to USD.

```python
df_usd = df.copy()
for col in ['Bet', 'Bonus', 'Profit']:
    df_usd[col] = df_usd[col] * BTC_USD_RATE

rainbow(df_usd.drop(columns=ID_COLS).describe())
```
